## DATA CLEANING AND NULL HANDLING 

```python
#Fill missing purchases with 0
df['purchased_last_month'] = df['purchased_last_month'].fillna(0)
```

```python
# fill missing categorical data with descriptive labels
df['sustainability_tags'] = df['sustainability_tags'].fillna('No Tag')
df['buy_box_availability'] = df['buy_box_availability'].fillna('Unknown')
```

```python
#  Drop Rows with Missing Critical Information
critical_cols = [
    'product_rating', 
    'total_reviews', 
    'discounted_price', 
    'original_price', 
    'discount_percentage',
    'product_page_url'
]

df = df.dropna(subset = critical_cols)
```

```python
# List the columns that are useless for visualization
useless_columns = [
    'product_image_url', 
    'product_page_url'
]
df = df.drop(columns=useless_columns)
```

```python
df.isnull().sum()
```

```python
# now convert to datetime
df['delivery_date'] = pd.to_datetime(df['delivery_date'], errors='coerce')
df['data_collected_at'] = pd.to_datetime(df['data_collected_at'], errors='coerce')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
#nulls date handling with collected date + 9 days
estimated_dates = df['data_collected_at'] + pd.Timedelta(days=9)
df['delivery_date'] = df['delivery_date'].fillna(estimated_dates)
```

```python
# Clean Text
text_cols = ['product_title', 'is_best_seller', 'is_sponsored', 'has_coupon', 'product_category']
for col in text_cols:
    if df[col].dtype == 'object':
        df[col] = df[col].str.strip()
```
